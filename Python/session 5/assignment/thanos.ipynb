```python
import random
import os

def make_dir(directory):
    '''Create file inside the directory given and if it exist do not create it.

    :param directory: path of file will be created inside it
    :return: 20 files inside the directory path  
    
    '''

    counter =0
    if not os.path.exists(directory):
        os.makedirs(directory)
    for i in range(20):
        dir =os.path.join(directory , "dir_"+str(i))
        if not os.path.exists(dir):
            os.makedirs(dir)
            counter +=1

    print(f"The number of files is {counter}")

    # remove 10 random files 
    for _ in range(10):
            dir = os.path.join(directory , "dir_"+str(random.randrange(0,20)))
            if os.path.exists(dir):
                os.rmdir(dir)
                counter -= 1

    print("Done")


        


path ="G:/DEPI/Giz4_AIS2_ML/Python/session 5/code/New_folder"
make_dir(path)
```
