```python
''' 
    reverse funtion
    param_1: receirve from the user text
    type : string

    return :revesed text
    type: string
'''

def reverse(s):
    my_list =s.split(" ")
    print(my_list)
    new_list = my_list[::-1]
    reversed = " ".join(new_list)
    print(reversed)
```

```python
''' 
generate random password function
param_1: take from the user the length of the password that want to generate
type: int
return: random password
type :string

'''

import random
def generate_passward(length):
    chars ="abcdefghijklmnopqrstuvwxyzABCDEFGHIJKLMNOPQRSTUVWXYZ1234567890!@#$%^&*"
    password =""
    for _ in range(length):
        password +=random.choice(chars)
    return password
```

```python
print(generate_passward(7))
print(generate_passward(11))
```

```python
words = ["Welcome", "Ali", "Hi", "Ali", "No", "Hi", "No", "Ali", "No", "Ali"]
```

```python

def count_dublicate(words):
    unique_words =set(words)
    count ={word:0 for word in unique_words}
    for i in unique_words:
        for j in words:
            if i == j:
                count[i] +=1
    return count

print(count_dublicate(words))
```

```python

import random 
responses = {
    "Hello" : ["Hello" , " Hi" ,"Greating"],
    "how are you?" : ["I am fine" ," I am great" , "I am doing well"],
    "goodbye" : ["Goodbye" , "See you later" ,"farewell"],
    "default" : ["I am sorry" ,"I did not understand" ,"could you please rephrase that?"]
}



def get_respons(user_input):
    for key in responses:
        if key in user_input:
            return random.choice(responses[key])
    return random.choice(responses["default"])



def chatbot():
    print("ChatBot: Hello , How can i help you?")
    while(True):
        user = input("User:    ").lower()
        respond = get_respons(user)
        print("chatbot: ",respond)
        if user == "goodbye":
            break
```

```python
chatbot()
```
