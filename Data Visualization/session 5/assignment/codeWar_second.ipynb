Fourth Problem

```python
class ListNode(object):
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next


class Solution(object):       
    def addTwoNumbers(self, l1, l2):

        dummy = ListNode(0)
        current = dummy
        remainder = 0

        while l1 or l2 or remainder:

            val1 = l1.val if l1 else 0
            val2 = l2.val if l2 else 0

            value = val1 + val2 + remainder

            remainder = value // 10
            digit = value % 10

            current.next = ListNode(digit)
            current = current.next

            if l1:
                l1 = l1.next
            if l2:
                l2 = l2.next

        return dummy.next
```

## Problem Six

```python
class Solution(object):
    def searchInsert(self, nums, target):
        """search the given number in the list and return the index

        :param nums: the list we search in it
        :param target: the integer search on at the list
        :return : the index of the number in the list 
        """
        l = 0
        r = len(nums) - 1
        while r>l :
            mid = l + (r-l)//2
            if nums[mid] == target:
                return mid
            elif nums[mid] > target:
                r = mid 
            elif nums[mid] < target:
                l = mid +1
        if nums[r] >= target:
            return r
        else:
            return r+1

sol =Solution()
print(sol.searchInsert([1,3,5,6],7))
```
