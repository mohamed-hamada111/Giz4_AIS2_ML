```python
def reverse_words(s):
    ''' Function used to reverse the given string
    
    :param str: the string to be reversed
    :return reversed: the reverse of the given string'''
    my_list = s.split(' ')
    result =' '.join(my_list[::-1])
    return result
s ='The greatest victory is that which requires no battle'
reverse_words(s)
```

```python
def are_you_playing_banjo(name):
    ''' a function which answers the question "Are you playing banjo?"
    
    :param name: the name of the person to be checked
    return: if he play banjo or not'''

    if name.startswith('r') or name.startswith('R'):
        return f'{name} plays banjo'
    else:
        return f'{name} does not play banjo'

are_you_playing_banjo('Ramy')
```

```python
def twoSum( nums, target):
        """Given an array of integers nums and an integer target, 
        return indices of the two numbers such that they add up to target.

        :param nums: List of numbers.
        :param target: the number we get its sum from nums
        :return: list with the indecies of the values their sum equal the target
        """
        result =[]
        flag = False
        for i in range(len(nums)-1):
                
                j =len(nums)-1
                while j > i:
                        if nums[i]+nums[j] == target:  
                                result.append(i)
                                result.append(j)
                                flag =True
                                break
                        j = j-1
                if flag ==True:
                        break
        return result
twoSum([3,2,4],6)
```

```python
def twoSum( nums, target):
        """Given an array of integers nums and an integer target, 
        return indices of the two numbers such that they add up to target.

        :param nums: List of numbers.
        :param target: the number we get its sum from nums
        :return: list with the indecies of the values their sum equal the target
        """
        num_map ={}
        for i ,num in enumerate(nums):
                value = target - num
                if value in num_map:
                        return [num_map[value],i]
                num_map[num]=i
        return []
                       
    
twoSum([3,2,4],6)
```
