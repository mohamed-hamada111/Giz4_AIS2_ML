```python
'''
    Task 1
'''
# Validate the Input
email ="Amit_ml@gmail.edu"
if((email.count('@') == 1 ) and (email[email.index('@'): ].count('.') > 0)):
    print("Valid Email")
else:
    print("Invalid Email")



```

```python
# Extract The Username
username = email[: email.index('@')]
print(username)

# Extract The Domain
Domain =email[email.index('@')+1 : email.rfind('.')]
print(Domain)
```

```python
# Check for Domain Ending
if email.endswith(".com"):
    print("Commercial Domain")
elif email.endswith(".edu"):
    print("Educational Domain")
else:
    print("Other Domain")
```
