```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.linear_model import LinearRegression ,Ridge ,Lasso
from sklearn.metrics import mean_squared_error ,r2_score
```

```python
x= np.array([[1],[2],[3],[4]])
y =np.array([2,4,6,8])
```

```python
lin_reg = LinearRegression()
lin_reg.fit(x,y)
y_pred = lin_reg.predict(x)

lin_reg.coef_
lin_reg.intercept_
```

```python
ridge = Ridge(alpha= 10.0)
ridge.fit(x,y)
y_pred_ridge = ridge.predict(x)
print(ridge.coef_)
ridge.intercept_
```

```python
lasso = Lasso(alpha= 10.0)
lasso.fit(x,y)
y_pred_lasso = lasso.predict(x)
print(lasso.coef_)
print(lasso.intercept_)
```

```python
print('Linear Regression: ')
print('MSE: ',mean_squared_error(y ,y_pred))
print('R2_Score: ',r2_score(y ,y_pred))
```

```python
print('Ridge Regression: ')
print('MSE: ',mean_squared_error(y ,y_pred_ridge))
print('R2_Score: ',r2_score(y ,y_pred_ridge))
```

```python
print('Lasso Regression: ')
print('MSE: ',mean_squared_error(y ,y_pred_lasso))
print('R2_Score: ',r2_score(y ,y_pred_lasso))
```

```python
plt.figure(figsize=(12,4))
plt.subplot(1,3,1)
plt.scatter(x ,y ,color= 'blue', label ='Actual Data')
plt.plot(x ,y_pred , color ='red' ,label ="linear regression")
plt.title("linear regression")
plt.legend()
plt.show()
```

```python
plt.figure(figsize=(12,4))
plt.subplot(1,3,1)
plt.scatter(x ,y ,color= 'blue', label ='Actual Data')
plt.plot(x ,y_pred_ridge , color ='red' ,label ="Ridge regression")
plt.title("ridge regression")
plt.legend()
plt.show()
```

```python
plt.figure(figsize=(12,4))
plt.subplot(1,3,1)
plt.scatter(x ,y ,color= 'blue', label ='Actual Data')
plt.plot(x ,y_pred , color ='red' ,label ="lasso regression")
plt.title("lasso regression")
plt.legend()
plt.show()
```

```python
plt.scatter(x,y ,color ='blue' , label ='Actual Data')
plt.plot(x,y_pred ,color ='red' ,label ='Linear Regression')
plt.plot(x,y_pred_ridge ,color ='green' , linestyle= '--' ,label ='Ridge Regression')
plt.plot(x,y_pred_lasso ,color ='purple',linestyle= ':'  ,label ='Lasso Regression')
plt.title("Comparesion of linear ,Ridge ,Lasso  Regression")
plt.xlabel('X')
plt.ylabel('Y')
plt.legend()
plt.show()
```
