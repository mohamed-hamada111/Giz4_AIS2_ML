```python
import numpy as np 
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.linear_model import LinearRegressionGD
```

```python
X = [50, 60, 70, 80, 90]
y = [150, 180, 210, 240, 270]

x_arr = np.array(X)
y_arr = np.array(y)
print(x_arr)
print(y_arr)
```

```python
class LinearRegression_Mohamed_Hamada:
    def __init__(self ,lr =0.001 ,iteration =100 ,w=1 ,b=1):
        self.lr = lr
        self.iteration = iteration
        self.w = w
        self.b = b
        self.losses =[]

    def fit(self,x,y):
        print(f"Theata_0 : \n w= {self.w} ,b= {self.b}")
        for _ in range(self.iteration):

            y_pred = self.w * x +self.b

            self.losses.append((1/x.shape[0])* np.sum((y_pred - y)**2))

            dw = (2/x.shape[0])* np.sum((y_pred - y)*x)
            db = (2/x.shape[0])* np.sum(y_pred - y)

            self.w = self.w - self.lr * dw
            self.b = self.b - self.lr *db
        
        print(f"Theata_1 : \n w= {self.w} ,b= {self.b}")

    def predict(self,x):
        return self.w *x +self.b
    
    def plot_training(self):
        
        plt.figure(figsize=(15,10))
        plt.plot(self.losses)
        plt.title('Model MSE')
        plt.show()
```

```python
model = LinearRegression_Mohamed_Hamada(0.001 , 100)
model.fit(x_arr,y_arr)
print(f"the predicted values is {model.predict(70)}")
model.plot_training()
```
