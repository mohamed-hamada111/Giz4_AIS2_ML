```python
def calculator(x ,y ,op ):
    if op == "+":
        return f"the first number is {x} , the second number is {y} , the result is {x+y}"
    elif op == "-":
        return f"the first number is {x} , the second number is {y} , the result is {x-y}"
    elif op == "*":
        return f"the first number is {x} , the second number is {y} , the result is {x*y}"
    elif op == "/":
        return f"the first number is {x} , the second number is {y} , the result is {x/y}"
    else :
        return "invalid operation"
    


```

```python
calculator(5,4,"+")
```

```python
def factorial(num):
    if num == 0 or num ==1:
        return num
    elif num <0:
        return 0
    return num* factorial(num-1)
```

```python
factorial(5)
```

```python
def isprime(num):
    for i in range(2,num):
        if (num % i) == 0:
            return False
    return True
```

```python
isprime(6)
```

```python
''' 
    number_1 : take integer number from the user 
    type : int
    number_2 : take the second integer number from the user 
    type : int
    return : a list contain the common divisor
    type : list
'''

def divisors(num1 ,num2):
    mn = min(num1 , num2)
    common_divisor =[]
    for i in range(1 ,mn+1):
        if num1 % i == 0 and num2 % i ==0:
            common_divisor.append(i)
    return common_divisor
```

```python
divisors( 8 ,12)
```
