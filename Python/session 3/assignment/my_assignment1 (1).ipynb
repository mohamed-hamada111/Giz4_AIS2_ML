# Write a Python program to removes multiple items from a given set

## Example
my_set: {1, 2, 3, 4, 5}

items to be removed: {1, 4}

Output: {2, 3, 5}

```python
# Write your answer here

''' 
    remove_set2 function 
    param_1: take from user set of numbers 
    type: set
    param_2: take from user the number want to delete it from set
    type: integer
    return new set without the number 
    type:set
'''
def remove_set2(my_set , num):
    while(num):
        my_set.remove(num.pop())
    return my_set

remove_set2({1,2,3,4,5} , {1,4})
```

# Take a string from the user and if it's Palindrome print "Palindrome" otherwise print "Not a Palindrome"

A string is Palindrome if it's read the same from the left and right

## Example
Input: racecar

Output: Palindrome

```python
# Write your answer here

''' 
palindrome Function
param_1: take word from the user
type: string
no return
'''
def palindrome(word):
    word_list = list(word)
    reverse = word_list[:: -1]
    if reverse == word_list:
        print("palindrome")
    else:
        print('Not a Parindrome')

palindrome('level')
```

# find the factorial of a number 
## Example
(factorial of 5) = 5 * 4 * 3 * 2 * 1 = 120

```python
# Write your answer here

''' 
    factorial Function
    param_1: take number from the user
    type: integer
    return the factorial of the number
    type: integer
'''
def factorial(num):
    if num <= 1:
        return 1
    return num *factorial(num-1)

factorial(5)
```

# Create a List of Squares of Numbers from 1 to 10
## Example
Output: [1, 4, 9, 16, 25, 36, 49, 64, 81, 100]

```python
# Write your answer here
''' 
    square function
    param_1: take number from the user
    type:integer
    return a list of all squares of the numbers from 1 to the number
    type: list
'''

def square(num):
    new_list=[]
    for i in range(1,num+1):
        new_list.append(i*i)
    return new_list

square(10)
```

# Create a function that Calculates the Sum of Even Numbers in a List
## Example
numbers: [1, 2, 3, 4, 5, 6]

Output: 12

```python
# Write your answer here

''' 
    even_sum Function
    param_1: take list from the user
    type: list
    return the summation of even numbers in that list 
    type: integer
'''
def even_sum(new_list):
    sum=0
    for i in new_list:
        if i %2 == 0:
            sum+=i
    return sum

even_sum([1,2,3,4,5,6])
```

# create a function that calculates the median of a list of numbers (a median is the middle element in sorted list of numbers)

## Example
numbers: [5, 2, 7, 3, 1]

Output: 3

```python
# Write your answer here

'''  
    median Function
    param_1: take a list from the user
    type: list
    return the median of the list
    type: integer
'''

def median(new_list):
    new_list.sort()
    index =round((len(new_list))/2)
    if index %2!=0:
        med =[new_list[index], new_list[index+1]]
    else:
        med = new_list[index]
    return med

newlist =[5,7 ,2,3,1]
median(newlist)
```

# create a function that calculates the mode of a list of number (mode is the most frequent element in a list)

## Example
lst: [5, 2, 7, 3, 2, 1, 5]

Output: 5

```python
# Write your answer here

'''  
    mode Function
    param_1: take list from the user 
    type: list
    return the most freqent element in a list
    type: integer
'''

def mode(new_list):
    dict ={}
    for i in new_list:
        dict.update({i:new_list.count(i)})
    print(dict)
    max=0
    for k,i in dict.items():
        if i > max: 
            max =k
    return max


list =[5,2,7,3,2,1,5]
mode(list)
```
