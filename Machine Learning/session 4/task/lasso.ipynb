```python
import numpy as np
import pandas as pd
```

```python
class LinearRegression_Mohamed_Hamada:
    def __init__(self ,lr =0.01 ,iteration =10 ,w=1 ,b=1):
        self.lr = lr
        self.iteration = iteration
        self.w = w
        self.b = b

    def fit(self,x,y):
        for _ in range(self.iteration):

            y_pred = self.w * x +self.b

            loss = (1/x.shape[0])* np.sum((y_pred - y)**2)

            dw = (2/x.shape[0])* np.sum((y_pred - y)*x)
            db = (2/x.shape[0])* np.sum(y_pred - y)

            self.w = self.w - self.lr * dw
            self.b = self.b - self.lr *db
        
        return self.w ,self.b 
    
    def lasso(self ,x ,y , alpha =0.1):
        for _ in range(self.iteration):
            y_pred = self.w *x +self.b
            loss =(1/x.shape[0]) *np.sum((y_pred -y)**2) +alpha * self.w
            dw = (2/x.shape[0])* np.sum((y_pred - y)*x) + alpha * np.sign(self.w)
            db = (2/x.shape[0])* np.sum(y_pred - y)
            self.w = self.w - self.lr *dw
            self.b = self.b - self.lr *db
        return self.w ,self.b
    
    def Ridge(self,x,y,alpha =0.1):
        for _ in range(self.iteration):
            y_pred =self.w*x +self.b
            loss = (1/x.shape[0])*np.sum((y_pred -y)**2)+ alpha * (self.w)**2
            dw = (2/x.shape[0]) * np.sum((y_pred - y)*x) + 2 * alpha * self.w
            db = (2/x.shape[0])* np.sum(y_pred - y)
            self.w = self.w - self.lr *dw
            self.b = self.b - self.lr * db

        return self.w ,self.b


    def predict(self,x):
        return self.w *x +self.b
```
